```python
from dbfread import DBF

# Load the DBF file into a DataFrame
table = DBF('inputs/roms/tide_data_east.dbf')
```

```python
print(table.field_names)
```

```python
import matplotlib.pyplot as plt

# 1. Extract only the WDEPTH column into a list
depths = [record['WDEPTH'] for record in table]

# 2. Plot the distribution
plt.figure(figsize=(10, 6))
plt.hist(depths, bins=50, color='skyblue', edgecolor='black')
plt.title('Distribution of Water Depth (WDEPTH)')
plt.xlabel('Depth (meters)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
